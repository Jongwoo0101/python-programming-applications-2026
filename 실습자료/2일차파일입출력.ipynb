## 파일 입출력 개념
- 키보드와 화면의 입출력 담당: input(), print()
- 파일에서 내용 읽어옴: read(), readline(), readlines()
- 파일에 내용을 씀: write(), writelines()
- 필요하다면 키보드에서 입력받은 후 파일에 쓸 수도 있고, 파일에서 내용을 입력받은 후 화면에 출력할 수도 있음

## 파일 입출력 과정
1. 파일 열기
2. 파일 읽기 및 파일 쓰기
3. 파일 닫기

## 1단계: 파일 열기
- open() 함수의 마지막 매개변수를 모드라고 함
- 파일을 열 때 어떤 용도로 여는지 결정해주는 것

|종류|설명|
|---|---|
|생략|r과 동일하다|
|r|읽기 모드, 기본값이다.|
|w|쓰기 모드, 기존에 파일이 있으면 덮어쓴다|
|r+|읽기/쓰기 겸용 모드|
|a|쓰기 모드. 기존에 파일이 있으면 이어서 쓴다. append의 약어|
|t|텍스트 모드. 텍스트 파일을 처리한다.|
|b|이진 모드. 이진파일을 처리한다|



## 2단계: 파일 처리
- 파일에 데이터를 쓰거나 파일에서 데이터를 읽어 올 수 있는 상태가 되었음

## 3단계: 파일 닫기
- 파일과 관련된 모든 작업이 끝나면 파일을 정상적으로 닫아야 함
- open() 함수로 연 변수명을 이용하여 파일을 닫음.  
```변수명.close()```

## 한 행씩 읽어 들이기

````python
inFp = None
inStr = ""

inFp = open("./TX/data1.txt", "r")

inStr = inFp.readline()
print(inStr, end="")

inStr = inFp.readline()
print(inStr, end="")

inStr = inFp.readline()
print(inStr, end="")

inFp.close()
````

````python
# 파일에 있는 모든 행을 읽기 위해 수정된 코드

inFp = None
inStr = ""

inFp = open("./TX/data1.txt")

while True :
    inStr = inFp.readline()
    if inStr == "" :
        break
    print(inStr, end="")
    
inFp.close()
````

## 한 번에 모두 읽어 들이기
- readline() 함수는 한 행씩 읽지만, readlines() 함수는 파일 내용을 통째로 읽어서 리스트에 저장함

````python
inFp = None
inList = ""

inFp = open("./TX/data1.txt")

inList = inFp.readlines()
print(inList)

inFp.close()
````

````python
# 한 행씩 출력하기
inFp = None
inList, inStr = [], ""

inFp = open("./TX/data1.txt")

inList = inFp.readlines()
for inStr in inList :
    print(inStr, end="")
    
inFp.close()
````

## 텍스트 파일을 이용한 출력
- 파일에 내용을 쓸 때는 write(), writelines() 함수를 사용하면 파일에 영구적으로 저장됨

## 한 행씩 파일에 쓰기
- 키보드에서 입력한 내용을 한 행씩 파일에 쓰는 코드
- 아무것도 입력하지 않은 채 엔터를 누르면 파일 쓰기를 종료

````python
outFp = None
outStr = ""

outFp = open("./TX/data2.txt", "w")

while True :
    outStr = input("내용 입력 : ")
    if outStr != "" :
        outFp.writelines(outStr + "\n")
    else :
        break
    
outFp.close()
print("---정상적으로 파일에 씀---")
````

## 이진 파일의 개념
- 이진(Binary) 파일은 글자가 아닌 비트(Bit) 단위로 의미가 있는 파일
- 그림 파일, 음악 파일, 동영상 파일, 엑셀 파일, 실행 exe 파일 등이 모두 이진 파일
- 텍스트 파일과 이진 파일을 구분하는 간단한 방법은 파일을 메모장에서 열어 보는 것

## 이진 파일 복사

````python
# 이진 파일을 복사해서 새로운 이진 파일로 만드는 코드
inFp, outFp = None, None
inStr = ""

inFp = open("./BIN/sample.bin", "rb")
outFp = open("./BIN/sample_copy.bin", "wb")

while True:
    inStr = inFp.read(1)
    if not inStr:
        break
    outFp.write(inStr)

inFp.close()
outFp.close()

print("--- 이진 파일이 정상적으로 복사되었음 ---")
````
