## 문자열 찾기

#### 문자열 메서드 한눈에 보기

| 메서드 | 기능 | 반환값 | 못 찾으면 |
|--------|------|--------|-----------|
| `count()` | 문자열 개수 | 개수(int) | `0` |
| `find()` | 첫 위치 찾기 | 인덱스 | `-1` |
| `rfind()` | 마지막 위치 찾기 | 인덱스 | `-1` |
| `index()` | 첫 위치 찾기 | 인덱스 | `ValueError` |
| `rindex()` | 마지막 위치 찾기 | 인덱스 | `ValueError` |
| `startswith()` | 시작 여부 확인 | `True` / `False` | `False` |
| `endswith()` | 끝 여부 확인 | `True` / `False` | `False` |

---

#### `count()`

```python
text = "banana"

print(text.count("a"))      # 3
print(text.count("na"))     # 2
print(text.count("x"))      # 0
```

---

#### `find()`

```python
text = "banana"

print(text.find("a"))       # 1
print(text.find("na"))      # 2
print(text.find("x"))       # -1
```

---

#### `rfind()`

```python
text = "banana"

print(text.rfind("a"))      # 5
print(text.rfind("na"))     # 4
print(text.rfind("x"))      # -1
```

---

#### `index()`

```python
text = "banana"

print(text.index("a"))      # 1
print(text.index("na"))     # 2

# print(text.index("x"))
# ValueError
```

---

#### `rindex()`

```python
text = "banana"

print(text.rindex("a"))     # 5
print(text.rindex("na"))    # 4

# print(text.rindex("x"))
# ValueError
```

---

#### `startswith()`

```python
text = "Python Programming"

print(text.startswith("Python"))   # True
print(text.startswith("Java"))     # False
```

---

#### `endswith()`

```python
text = "Python Programming"

print(text.endswith("Programming"))   # True
print(text.endswith("Python"))        # False
```

---

#### 비교

| 비교 | 차이 |
|------|------|
| `find()` vs `index()` | `find()` → `-1`, `index()` → `ValueError` |
| `rfind()` vs `rindex()` | `rfind()` → `-1`, `rindex()` → `ValueError` |
| `startswith()` | 시작 여부 확인 |
| `endswith()` | 끝 여부 확인 |

---

#### 핵심 암기

| 메서드 | 기억할 것 |
|--------|-----------|
| `count()` | 개수 반환 |
| `find()` | 첫 위치, 없으면 `-1` |
| `rfind()` | 마지막 위치, 없으면 `-1` |
| `index()` | 첫 위치, 없으면 오류 |
| `rindex()` | 마지막 위치, 없으면 오류 |
| `startswith()` | 시작 여부 |
| `endswith()` | 끝 여부 |

> **암기 팁**
> - `find` → `-1`
> - `index` → `ValueError`
> - `r` → 오른쪽(뒤쪽)부터 검색
> - `start` → 시작 확인
> - `end` → 끝 확인

````python
ss = '파이썬 공부는 즐겁습니다. 물론 모든 공부가 다 재미있지는 않죠. ^^'

print(ss.count('공부'))
print(ss.find('공부'), ss.rfind('공부'), ss.find('공부', 5), ss.find('없다'))
print(ss.index('공부'), ss.rindex('공부'), ss.index('공부', 5))
print(ss.startswith('파이썬'), ss.startswith('파이썬', 10), ss.endswith('^^'))
````

## 문자열 공백 삭제/변경하기

#### 문자열 메서드 (`strip`, `rstrip`, `lstrip`, `replace`)

| 메서드 | 기능 | 반환값 |
|--------|------|--------|
| `strip()` | 양쪽 공백(또는 지정 문자) 제거 | 새로운 문자열 |
| `lstrip()` | 왼쪽 공백(또는 지정 문자) 제거 | 새로운 문자열 |
| `rstrip()` | 오른쪽 공백(또는 지정 문자) 제거 | 새로운 문자열 |
| `replace()` | 문자열 변경 | 새로운 문자열 |

---

#### `strip()`

양쪽의 공백을 제거합니다.

```python
text = "   Python   "

print(text.strip())
```

**결과**

```python
Python
```

지정한 문자도 제거할 수 있습니다.

```python
text = "###Python###"

print(text.strip("#"))
```

**결과**

```python
Python
```

---

#### `lstrip()`

왼쪽 공백만 제거합니다.

```python
text = "   Python   "

print(text.lstrip())
```

**결과**

```python
Python   
```

지정한 문자도 제거할 수 있습니다.

```python
text = "###Python###"

print(text.lstrip("#"))
```

**결과**

```python
Python###
```

---

#### `rstrip()`

오른쪽 공백만 제거합니다.

```python
text = "   Python   "

print(text.rstrip())
```

**결과**

```python
   Python
```

지정한 문자도 제거할 수 있습니다.

```python
text = "###Python###"

print(text.rstrip("#"))
```

**결과**

```python
###Python
```

---

#### `replace()`

특정 문자열을 다른 문자열로 변경합니다.

```python
text = "I like Python"

print(text.replace("Python", "Java"))
```

**결과**

```python
I like Java
```

모든 문자열을 변경합니다.

```python
text = "banana"

print(text.replace("a", "*"))
```

**결과**

```python
b*n*n*
```

변경 횟수를 지정할 수도 있습니다.

```python
text = "banana"

print(text.replace("a", "*", 2))
```

**결과**

```python
b*n*na
```

---

#### 비교표

| 메서드 | 제거 위치 |
|--------|----------|
| `strip()` | 양쪽 |
| `lstrip()` | 왼쪽 |
| `rstrip()` | 오른쪽 |
| `replace()` | 문자열 변경 |

---

#### 핵심 암기

| 메서드 | 기억할 것 |
|--------|-----------|
| `strip()` | 양쪽 제거 |
| `lstrip()` | 왼쪽 제거 |
| `rstrip()` | 오른쪽 제거 |
| `replace()` | 문자열 변경 |

> **암기 팁**
> - **lstrip** → Left(왼쪽)
> - **rstrip** → Right(오른쪽)
> - **strip** → 양쪽 제거
> - **replace** → 문자열 바꾸기

````python
ss = ' 파 이 썬 '
print(ss.strip())
print(ss.rsplit())
print(ss.lstrip())

ss = '열심히 파이썬 공부 중~~'
print(ss.replace('파이썬', 'Python'))
````

````python
ss = '---파---이---썬---'
print(ss.strip('-'))
ss = '<<<파<<이>>썬>>>'
print(ss.strip('<>'))
````

## 문자열 중간의 공백까지 삭제해주는 코드

````python
inStr = "   한글 Python 프로그래밍    "
outStr = ""

for i in range(0, len(inStr)) :
    if inStr[i] != ' ':
        outStr += inStr[i]
        
print("원래 문자열 ==> " + '[' + inStr + ']')
print("공백 삭제 문자열 ==> " + '[' + outStr + ']')
````

## 문자열을 입력받아 그 중 o를 $로 변경하는 문자열 변경을 응용

````python
ss = input("입력 문자열 ==> ")

print("출력 문자열 ==>", end='')
for i in range(0, len(ss)) :
    if ss[i] != 'o' :
        print(ss[i], end='')
    else :
        print('$', end='')
        
print(f"\n한 줄로 처리한 결과: {ss.replace('o', '$')}")
````

## 문자열 분리 결합치기

#### 문자열 메서드 한눈에 보기

| 메서드 | 기능 | 반환값 |
|--------|------|--------|
| `split()` | 문자열 분리 | 리스트(`list`) |
| `splitlines()` | 줄바꿈 기준 분리 | 리스트(`list`) |
| `join()` | 문자열 합치기 | 새로운 문자열(`str`) |

---

#### `split()`

구분자를 기준으로 문자열을 분리합니다.

```python
text = "apple,banana,grape"

print(text.split(","))
```

**결과**

```python
['apple', 'banana', 'grape']
```

공백 기준으로 분리

```python
text = "Python is easy"

print(text.split())
```

**결과**

```python
['Python', 'is', 'easy']
```

---

#### `splitlines()`

줄바꿈(`\n`)을 기준으로 문자열을 분리합니다.

```python
text = "Python\nJava\nC"

print(text.splitlines())
```

**결과**

```python
['Python', 'Java', 'C']
```

---

#### `join()`

리스트의 문자열을 하나의 문자열로 합칩니다.

```python
fruit = ["apple", "banana", "grape"]

print(",".join(fruit))
```

**결과**

```python
apple,banana,grape
```

공백으로 합치기

```python
words = ["Python", "is", "easy"]

print(" ".join(words))
```

**결과**

```python
Python is easy
```

---

#### 비교

| 메서드 | 기능 |
|--------|------|
| `split()` | 문자열 → 리스트 |
| `splitlines()` | 줄바꿈 기준 문자열 → 리스트 |
| `join()` | 리스트 → 문자열 |

---

#### 핵심 암기

| 메서드 | 기억할 것 |
|--------|-----------|
| `split()` | 문자열 분리 |
| `splitlines()` | 줄바꿈 기준 분리 |
| `join()` | 리스트를 문자열로 합치기 |

> **암기 팁**
> - `split` → **문자열 → 리스트**
> - `join` → **리스트 → 문자열**
> - `splitlines` → **줄바꿈(`\n`) 기준 분리**

````python
ss = 'Python을 열심히 공부 중'
print(ss.split())

ss = '하나:둘:셋'
print(ss.split(':'))

ss = '하나\n둘\n셋'
print(ss.splitlines())

ss = '%'
print(ss.join('파이썬'))
````

## 문자열 정렬/채우기

#### 문자열 메서드 한눈에 보기

| 메서드 | 기능 | 반환값 |
|--------|------|--------|
| `center()` | 가운데 정렬 | 새로운 문자열 |
| `ljust()` | 왼쪽 정렬 | 새로운 문자열 |
| `rjust()` | 오른쪽 정렬 | 새로운 문자열 |
| `zfill()` | 앞을 `0`으로 채움 | 새로운 문자열 |

---

#### `center()`

지정한 길이만큼 **가운데 정렬**합니다.

```python
text = "Python"

print(text.center(10))
print(text.center(10, "*"))
```

**결과**

```python
  Python  
**Python**
```

---

#### `ljust()`

지정한 길이만큼 **왼쪽 정렬**합니다.

```python
text = "Python"

print(text.ljust(10))
print(text.ljust(10, "-"))
```

**결과**

```python
Python    
Python----
```

---

#### `rjust()`

지정한 길이만큼 **오른쪽 정렬**합니다.

```python
text = "Python"

print(text.rjust(10))
print(text.rjust(10, "-"))
```

**결과**

```python
    Python
----Python
```

---

#### `zfill()`

앞쪽을 `0`으로 채웁니다.

```python
print("25".zfill(5))
print("123".zfill(6))
```

**결과**

```python
00025
000123
```

---

#### 비교

| 메서드 | 기능 |
|--------|------|
| `center()` | 가운데 정렬 |
| `ljust()` | 왼쪽 정렬 |
| `rjust()` | 오른쪽 정렬 |
| `zfill()` | 앞을 `0`으로 채움 |

---

#### 핵심 암기

| 메서드 | 기억할 것 |
|--------|-----------|
| `center()` | 가운데 정렬 |
| `ljust()` | 왼쪽 정렬 |
| `rjust()` | 오른쪽 정렬 |
| `zfill()` | 앞을 `0`으로 채움 |

> **암기 팁**
> - `center` → 가운데
> - `ljust` → Left(왼쪽)
> - `rjust` → Right(오른쪽)
> - `zfill` → Zero(0) 채우기

````python
ss = '파이썬'

print(ss.center(10))
print(ss.center(10, '-'))
print(ss.ljust(10))
print(ss.rjust(10))
print(ss.zfill(10))
````
